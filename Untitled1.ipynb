### Read TLE Files

```python
def read_tle_file(filepath):
    with open(filepath, 'r') as file:
        lines = file.readlines()
    return [line.strip() for line in lines if line.strip() != ""]

active_data = read_tle_file("active-sate.txt")
debris1     = read_tle_file("cosmos-1408-debris.txt")
debris2     = read_tle_file("cosmos-2251-debris.txt")
debris3     = read_tle_file("iridium-33-debris.txt")
debris4     = read_tle_file("fengyun-1c-debris.txt")

print("Files loaded successfully!")
print("Active satellite lines:", len(active_data))
print("Debris 1 lines:", len(debris1))
print("Debris 2 lines:", len(debris2))
print("Debris 3 lines:", len(debris3))
print("Debris 4 lines:", len(debris4))
```

### Parse TLE Data (Extract Numbers)

```python
def parse_tle(lines, label):
    records = []
    for i in range(0, len(lines) - 2, 3):
        try:
            line1 = lines[i + 1]
            line2 = lines[i + 2]

            if not line1.startswith('1') or not line2.startswith('2'):
                continue

            # Extract orbital parameters
            inclination  = float(line2[8:16])
            raan         = float(line2[17:25])
            eccentricity = float("0." + line2[26:33])
            arg_perigee  = float(line2[34:42])
            mean_anomaly = float(line2[43:51])
            mean_motion  = float(line2[52:63])
            bstar        = float(line1[53:61].replace(" ", ""))

            # Calculate extra features
            mu              = 398600.4418           # Earth gravity constant
            n               = mean_motion * 2 * np.pi / 86400
            semi_major_axis = (mu / n**2) ** (1/3)  # km
            velocity        = np.sqrt(mu / semi_major_axis)  # km/s
            altitude        = semi_major_axis - 6371          # km
            orbital_period  = 1440 / mean_motion              # minutes

            records.append({
                "inclination"    : inclination,
                "raan"           : raan,
                "eccentricity"   : eccentricity,
                "arg_perigee"    : arg_perigee,
                "mean_anomaly"   : mean_anomaly,
                "mean_motion"    : mean_motion,
                "bstar"          : bstar,
                "semi_major_axis": semi_major_axis,
                "velocity"       : velocity,
                "altitude"       : altitude,
                "orbital_period" : orbital_period,
                "label"          : label   # 0 = active, 1 = debris
            })
        except:
            continue
    return records

print(" Parser function ready")
```

###  Build the Dataset

```python
all_records  = parse_tle(active_data, label=0)  # active = 0
all_records += parse_tle(debris1,     label=1)  # debris = 1
all_records += parse_tle(debris2,     label=1)
all_records += parse_tle(debris3,     label=1)
all_records += parse_tle(debris4,     label=1)

df = pd.DataFrame(all_records)

print(" Dataset created")
print("Total rows :", df.shape[0])
print("Total columns:", df.shape[1])

print("\nActive vs Debris count:")
print(df["label"].value_counts().rename({0: "Active Satellites", 1: "Debris"}))

print("\nSample Data:")
print(df.head())
```
